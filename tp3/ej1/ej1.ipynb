- <span style="color:white">Existe una relacion entre el uso de dispositivos tecnologicos y el peso?</span>

<p>No existe, podemos ver como la correlacion nos da un numero bajo donde hay no relacion alguna -> Falso.



```python
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
var_1 = 'FAF'
var_2 = 'Weight'

#creamos el scatter plot
plt.figure(figsize=(6, 6))
plt.scatter(archivoCSV[var_1], archivoCSV[var_2])
plt.xlabel(var_1)
plt.ylabel(var_2)
plt.title("Relación entre {} y {}".format(var_1, var_2))

# calcular la correlación de Pearson usando SciPy
coeficiente_correlacion, _ = stats.pearsonr(archivoCSV[var_1], archivoCSV[var_2])
# agregar la leyenda con el valor del coeficiente de correlación
plt.legend([f"Correlación: {coeficiente_correlacion:.4f}"], loc="upper left")

# Mostrar el gráfico
plt.grid(True)
plt.show()
```

- <span style="color:white">Existe una relacion lineal inversa entre la frecuencia de actividad fisica y el peso</span>
