**Que representan?**

- $\color{orange}{\text{age}}$: Edad.

- $\color{orange}{\text{Height}}$: Altura.
- $\color{orange}{\text{Weight}}$: Peso.
- $\color{orange}{\text{family\_history\_with\_overweight}}$: Tiene algun familiar que sufre o sufrio sobrepeso? - `Si o no`.
- $\color{orange}{\text{FAVC}}$: Frecuentemente consume comidas con altas calorias? - `Si o no`.
- $\color{orange}{\text{FCVC}}$: Consume frecuentemente vegetales? - `Nunca, a veces, siempre`.
- $\color{orange}{\text{NCP}}$: Numero de comidas principales?
- $\color{orange}{\text{CAEC}}$: Consumo de alimentos entre comidas - `No, a veces, frecuentemente, siempre`.
- $\color{orange}{\text{SMOKE}}$: Fuma? - `Si o no`.
- $\color{orange}{\text{CH2O}}$: Consumeo de agua diario - `Menos que un litro, entre 1 y 2 litros, mas de 2 litros`.
- $\color{orange}{\text{SCC}}$: Controla el consumo de calorias? - `Si o no`.
- $\color{orange}{\text{FAF}}$: Frecuencia de actividad fisica - `0, 1 a 2, 2 a 4, 4 a 5`.
- $\color{orange}{\text{TUE}}$: Tiempo usando dispositivos tecnologicos? - `0 a 2, 3 a 5, mayor a 5`
- $\color{orange}{\text{CALC}}$: Consumo de alcohol? - `no, a veces, frecuentemente, siempre`.
- $\color{orange}{\text{MTRANS}}$: Metodo de transporte usado - `auto, moto, bici, transporte publico, caminata`.
- $\color{orange}{\text{NObeyesdad}}$: Tipo de obesidad - `peso insuficiente, peso normal, sobrepeso nivel i, sobrepeso nivel 2, obsesidad tipo i, obesidad tipo ii, obesidad tipo iii`.
- $\color{orange}{\text{BMI}}$: índice de masa corporal.
