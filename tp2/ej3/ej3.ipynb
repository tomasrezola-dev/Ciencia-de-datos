<p>1. Cuales son las peliculas con mejor proporcion imdbScore con imdbVotes

<p>Para eso usamos

- Escala logaritmica: Damos validez a como la cantidad de votantes se incrementa de a numero de ceros.

- Reescalado a [0, 1]: $$ \left( \frac{\log_{10}(v) - \log_{10}(v_{min})}{\log_{10}(v_{max}) - \log_{10}(v_{min})} \right) $$

- Reescalado a [0, 10]: $$ \textbf{votos\_score} = 10 \cdot \left( \frac{\log_{10}(v) - \log_{10}(v_{min})}{\log_{10}(v_{max}) - \log_{10}(v_{min})} \right) $$

- Ponderamos: 
    $$ (0.7 \cdot \text{imdbScore}) + (0.3 \cdot \text{votes\_scaled\_1\_10}) $$

```python
import numpy as np
log_votos = np.log10(df['imdbVotes']+1)
min_log = log_votos.min()
max_log = log_votos.max()
df['votes_scaled'] = 10 * ((log_votos - min_log) / (max_log - min_log))
df['rnk_rating'] = (0.7 * df['imdbScore']) + (0.3 * df['votes_scaled'])

df.sort_values(by='rnk_rating', ascending=False).head(3)
```

2. Cuales son las plataformas, que en promedio, tienen la mejor rnk_rating?

```python
grouped_data = df.groupby('platform')
media_platform = grouped_data['rnk_rating'].median()
print(media_platform.sort_values(ascending=False))
#media_platfrom es una serie (key, valor): (platform, media_rnk_rating)
```
