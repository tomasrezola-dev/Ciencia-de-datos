```python
import utils
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv("ObesityDataSet_raw_and_data_sinthetic.csv")
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df.select_dtypes(include="number"))
```

<h2>T-SNE

```python
from sklearn.manifold import TSNE
```

```python
x_tsne = TSNE(n_components=2, random_state=10,perplexity=30).fit_transform(df_scaled)
```

```python
# imprimimos los tamaños antes y después
print('Before t-SNE: {}'.format(df_scaled.shape))
print('After t-SNE: {}'.format(x_tsne.shape))

#graficamos cada punto mediante un scatter plot, donde cada punto es un registro del dataset
plt.scatter(x_tsne[:,0], x_tsne[:,1], c=df["Weight"], cmap='viridis')
plt.grid(True)
plt.xlabel('Proyección t-SNE 1')
plt.ylabel('Proyección t-SNE 2')
plt.title('Proyección del conjunto de datos a 2 dimensiones')
plt.colorbar()
plt.show()
```

```python
df_pca = pd.DataFrame({
    "PC1": x_tsne[:, 0],
    "PC2": x_tsne[:, 1],
    "Gender": df["Gender"].values
})

sns.scatterplot(
    data=df_pca,
    x="PC1",
    y="PC2",
    hue="Gender",
    palette=["pink", "blue"]
)
plt.title("Proyección del conjunto de datos a 2 dimensiones por género")
plt.show()
```

Podemos, comparandolo con el del inciso 1 c, observar como se agruparon grupos en las partes superiores de ambas componentes donde esa comparacion parece estar justificada por el sexo femenino

```python
df_pca = pd.DataFrame({
    "PC1": x_tsne[:, 0],
    "PC2": x_tsne[:, 1],
    "Comida con altas calorias": df["FAVC"].values
})

sns.scatterplot(
    data=df_pca,
    x="PC1",
    y="PC2",
    hue="Comida con altas calorias",
    palette=["#C34E4E", "#8AD981"]
)
plt.title("Proyección del conjunto de datos a 2 dimensiones por FAVC")
plt.show()
```

```python
df_pca = pd.DataFrame({
    "PC1": x_tsne[:, 0],
    "PC2": x_tsne[:, 1],
    "Monitoreo": df["SCC"].values
})

sns.scatterplot(
    data=df_pca,
    x="PC1",
    y="PC2",
    hue="Monitoreo",
    palette=["#CE3838", "#27DA30"]
)

plt.show()
```

```python
df_pca = pd.DataFrame({
    "PC1": x_tsne[:, 0],
    "PC2": x_tsne[:, 1],
    "Historia_fam_peso": df["family_history_with_overweight"].values
})

sns.scatterplot(
    data=df_pca,
    x="PC1",
    y="PC2",
    hue="Historia_fam_peso",
    palette=["#CE3838", "#27DA30"]
)

plt.show()
```

```python
df_pca = pd.DataFrame({
    "PC1": x_tsne[:, 0],
    "PC2": x_tsne[:, 1],
    "Monitoreo": df["CAEC"].values
})

sns.scatterplot(
    data=df_pca,
    x="PC1",
    y="PC2",
    hue="Monitoreo",
    palette=["#DEDE22", "#7EDE91", "#2D29BB", "#CE3838"]
)

plt.show()
```

<H2>UMAP

```python
from umap import UMAP
```

```python
x_map = UMAP(n_components=2).fit_transform(df_scaled)
```

```python
# imprimimos los tamaños antes y después
print('Antes de UMAP: {}'.format(df_scaled.shape))
print('Después de UMAP: {}'.format(x_map.shape))

#graficamos cada punto mediante un scatter plot, donde cada punto es un registro del dataset
plt.scatter(x_map[:,0], x_map[:,1], c=df["Weight"], cmap='viridis')
plt.grid(True)
plt.xlabel('Proyección UMAP 1')
plt.ylabel('Proyección UMAP 2')
plt.title('Proyección del conjunto de datos a 2 dimensiones')
plt.colorbar()
plt.show()
```
